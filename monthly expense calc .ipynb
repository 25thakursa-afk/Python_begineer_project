# ***Monthly expense calc***

```python
## room rent
# tiffin
# electricity
# total monthly expense

room = int(input("Enter monthly room rent= "))
tif = int(input("Enter total no. of tiffin= "))
elec = int(input("Enter monthly electricity= "))
exp = int(input("Enter monthly expense= "))

tiff = tif*60
cal = room + tiff + elec + exp
print("Your total monthly expense = ",cal)
```
